# AnyCompany revenue analysis

Read the `orders_enriched` table your pipeline produced and chart revenue by supplier. Run each cell in order, from top to bottom.

```python_glue_session
%idle_timeout 30
%glue_version 5.1
%worker_type G.1X
%number_of_workers 2

import sys
from awsglue.transforms import *
from awsglue.utils import getResolvedOptions
from pyspark.context import SparkContext
from awsglue.context import GlueContext
from awsglue.job import Job

sc = SparkContext.getOrCreate()
glueContext = GlueContext(sc)
spark = glueContext.spark_session
job = Job(glueContext)
```

```python_glue_session
import matplotlib.pyplot as plt

df = spark.sql("""
    SELECT supplier, SUM(quantity * unit_price) AS revenue
    FROM glue_lab_db.orders_enriched
    GROUP BY supplier
    ORDER BY revenue DESC
""")
pdf = df.toPandas()
pdf
```

```python_glue_session
ax = pdf.plot(kind="bar", x="supplier", y="revenue", legend=False, figsize=(5, 5))
ax.set_title("AnyCompany revenue by supplier")
ax.set_ylabel("Revenue (USD)")
ax.set_xlabel("Supplier")
plt.tight_layout()
%matplot plt
```
